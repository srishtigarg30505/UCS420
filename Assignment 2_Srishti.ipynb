3.Random






i)

```python
import random

random.seed(1024170451)
numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

print(numbers)
```

ii)

```python
odd = []

for n in numbers:
    if n % 2 != 0:
        odd.append(n)

print("Odd numbers:", odd)
print("Count of odd numbers:", len(odd))
```

iii)

```python
even = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)

print("Even numbers:", even)
print("Count of even numbers:", len(even))
```

iv)

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
```

```python
prime_numbers = [n for n in numbers if is_prime(n)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

v)

```python
most = numbers[0]
max_count = numbers.count(most)

for n in numbers:
    count = numbers.count(n)

    if count > max_count:
        most = n
        max_count = count

print("Most frequent number:", most)
print("Number of times it occurs:", max_count)
```

4.Sets


```python
digits = [1, 0, 2, 4, 1, 7, 0, 4, 5, 1]

A = set()
B = set()

for digit in digits:
    A.add(digit * 7)
    B.add(digit * 9)

print("Set A:", A)
print("Set B:", B)
```

i)


```python
union = A.union(B)

print("Union:", union)
```

ii)

```python
intersection = A.intersection(B)

print("Intersection:", intersection)
```

iii)

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference():  gives the values belonging to one set only,
# symmetric_difference():  gives values belonging to either set but not both.
```

iv)

```python
symmetric_difference = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_difference)
```

v)

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

vi)

```python
X = int(input("Enter a value: "))

A.discard(X)

print("Set A after removing X:", A)

# discard() is safer because it does not give an error if X is not present.
```

5.Dictionary


```python
my_dict = {
    "name": "Srishti",
    "roll_no": "1024170451",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

print(my_dict)
```

i)


```python
my_dict["location"] = my_dict.pop("city")

print(my_dict)
```

ii)

```python
my_dict["cgpa"] = 9.30

print(my_dict)
```

iii)

```python
my_dict["age"] = my_dict["age"] + 1

print(my_dict)
```

iv)

```python
dict_pop = my_dict.copy()

removed_value = dict_pop.pop("branch")

print(dict_pop)
print("Removed value:", removed_value)
```

```python
dict_del = my_dict.copy()

del dict_del["branch"]

print(dict_del)
```

v)

```python
for key, value in my_dict.items():
    print(key, ":", value)
```

vi)

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")
```

vii)

```python
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170452",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print(merged_dict)

# When both dictionaries have the same key, the value from the second dictionary wins.
```

viii)

```python
string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}

print(string_dict)
```
