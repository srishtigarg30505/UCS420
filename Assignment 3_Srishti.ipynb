```python
emp_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

df_emp = pd.DataFrame(emp_data)
df_emp.to_csv('employees.csv', index=False)
display(df_emp)
```

a) to e)

```python
# a)
print("a) Shape of DataFrame:", df_emp.shape)

# b)
print("\nb) Summary of DataFrame:")
df_emp.info()

# c)
print("\nc) Descriptive statistics:")
display(df_emp.describe())

# d)
print("\nd) First 5 rows:")
display(df_emp.head(5))
print("\nLast 3 rows:")
display(df_emp.tail(3))

# e)
print("\ne) Statistics:")
print("i. Average salary:", df_emp['Salary'].mean())
print("ii. Total bonus paid:", df_emp['Bonus'].sum())
print("iii. Youngest employee's age:", df_emp['Age'].min())
print("iv. Highest performance rating:", df_emp['Rating'].max())
```

f) to l)

```python
# f)
df_sorted = df_emp.sort_values(by='Salary', ascending=False)
print("f) Sorted by Salary (Descending):")
display(df_sorted)

# g)
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df_emp['Rating_Category'] = df_emp['Rating'].apply(categorize_rating)
print("\ng) Added Rating_Category:")
display(df_emp[['Name', 'Rating', 'Rating_Category']])

# h)
print("\nh) Missing values in DataFrame:")
print(df_emp.isnull().sum())

# i)
df_emp.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("\ni) Renamed column Employee_ID to ID:")
display(df_emp.head(2))

# j)
# i.
exp_gt_5 = df_emp[df_emp['Years_of_Experience'] > 5]
print("\nj) i. Employees with > 5 years experience:")
display(exp_gt_5)

# ii.
it_dept = df_emp[df_emp['Department'] == 'IT']
print("\nj) ii. Employees in IT department:")
display(it_dept)

# k)
df_emp['Tax'] = df_emp['Salary'] * 0.10
print("\nk) Added Tax column (10% of Salary):")
display(df_emp[['Name', 'Salary', 'Tax']])

# l)
df_emp.to_csv('modified_employees.csv', index=False)
print("\nl) Saved the modified DataFrame to 'modified_employees.csv'")
```
