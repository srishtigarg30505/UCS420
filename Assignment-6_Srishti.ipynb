**e.**
Vectorization lets numpy do the looping internally in C instead of python so its way faster. Boolean indexing is similar - instead of writing a loop to check each element one by one, numpy does it all at once on the whole array. Both of these avoid the python for loop overhead and work on contiguous memory which is more cache friendly.

**Q4.**

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("a. shape:", X.shape, "ndim:", X.ndim)

xt = X.T
print("\nb. X transposed:\n", xt)
```

**b. Transpose explanation:** X.T just flips rows and columns. So now rows become features (sleep, activity, stress) and columns become the users. Basically the matrix gets mirrored along the diagonal.

```python
xtx = X.T @ X
print("c. X.T @ X:\n", xtx)

inv = np.linalg.inv(xtx)
print("\nd. Inverse:\n", inv)

b = inv @ X.T @ y
print("\ne. Beta coefficients:", b)
```

**f.** The beta values tell us how much each feature contributes to the assistance score. b[0] is for sleep hours, b[1] is for activity level, b[2] is for stress. So if stress coeff is positive it means higher stress = higher assistance needed which makes sense. Each coefficient shows how much the score changes when that feature goes up by 1 unit (keeping others fixed).

```python
new = np.array([5, 40, 7])
pred = new @ b
print("g. Predicted score:", pred)
```
